1. Write Python scripts for basic file operations and data processing?
```
import os
import csv
from statistics import mean

# Create a sample CSV file
filename = "data.csv"
rows = [
    ["name", "age", "score"],
    ["Alice", "23", "88"],
    ["Bob", "21", "75"],
    ["Charlie", "22", "93"]
]

with open(filename, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(rows)

print(f"File '{filename}' created.")

# Read the CSV file
data = []
with open(filename, "r", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        data.append(row)

print("Read data:", data)

# Basic data processing: compute average age and score
ages = [int(row["age"]) for row in data]
scores = [int(row["score"]) for row in data]

avg_age = mean(ages)
avg_score = mean(scores)

print("Average age:", avg_age)
print("Average score:", avg_score)

# Filter and write processed data to a new file
filtered = [row for row in data if int(row["score"]) >= 80]

output_file = "high_scorers.csv"
with open(output_file, "w", newline="", encoding="utf-8") as f:
    fieldnames = ["name", "age", "score"]
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(filtered)

print(f"Filtered data written to '{output_file}'.")
```
2. Develop a simple web scraper to extract data from a website?
```
import requests
from bs4 import BeautifulSoup

# Target URL
url = "https://quotes.toscrape.com/"

response = requests.get(url)
if response.status_code != 200:
    raise SystemExit(f"Failed to fetch page: {response.status_code}")

soup = BeautifulSoup(response.text, "html.parser")

# Extract all quotes and authors from the page
quotes_data = []
quote_blocks = soup.select("div.quote")

for block in quote_blocks:
    text = block.select_one("span.text").get_text(strip=True)
    author = block.select_one("small.author").get_text(strip=True)
    tags = [tag.get_text(strip=True) for tag in block.select("div.tags a.tag")]
    quotes_data.append({"text": text, "author": author, "tags": tags})

# Display scraped data
for q in quotes_data:
    print(f"Quote: {q['text']}")
    print(f"Author: {q['author']}")
    print(f"Tags: {', '.join(q['tags'])}")
    print("-" * 40)
```
