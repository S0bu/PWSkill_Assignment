1. Set up an environment for integration testing and write test cases that validate the interaction between different components of a system?
```
For integration testing, I would create a separate test environment using pytest as the test framework and Docker to ensure the application runs in a clean, reproducible containerized setup. The idea is to validate how different parts of the calculator system work together, such as the application logic in calculator.py, the API layer or CLI interface, and the runtime environment inside the container.

The environment would include the application code, a requirements.txt file with pytest and app dependencies, a Dockerfile to build the test image, and optionally a docker-compose.yml file if multiple services were involved. Since a calculator app is small, one container is enough, but the same setup can scale to larger systems where multiple components must interact.

Project structure

*calculator.py — contains calculator functions.

*test_calculator_integration.py — contains pytest integration tests.

*requirements.txt — includes pytest.

*Dockerfile — builds the containerized test environment.
```

```
calculator.py:
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        raise ValueError("Cannot divide by zero")
    return a / b
```
```
Dockerfile:
FROM python:3.11-slim

WORKDIR /app

COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

COPY . .

CMD ["pytest", "-v"]
```
```
test_calculator_integration.py:
import pytest
from calculator import add, subtract, multiply, divide

def test_addition_integration():
    result = add(10, 5)
    assert result == 15

def test_subtraction_integration():
    result = subtract(10, 5)
    assert result == 5

def test_multiplication_integration():
    result = multiply(4, 3)
    assert result == 12

def test_division_integration():
    result = divide(20, 4)
    assert result == 5

def test_division_by_zero_integration():
    with pytest.raises(ValueError, match="Cannot divide by zero"):
        divide(10, 0)

def test_chained_operations_integration():
    result = divide(multiply(add(2, 3), 4), 2)
    assert result == 10
```
```
Build and run the tests with Docker:
docker build -t calculator-test .
docker run --rm calculator-test
```
