#Linux

Q1. What is Linux?
- Linux is an open-source, Unix-like operating system kernel that powers various operating systems known as Linux distributions (distros). It is widely used for servers, desktops, mobile devices, and embedded systems. Linux is known for its stability, security, and flexibility.

Q2. What is the difference between Hard Link & Soft Link?
- A Hard Link is an additional name for an existing file on the file system; both the original and the hard link point to the same inode (data). Deleting either does not remove the file data.
- A Soft Link (or symbolic link) is a pointer to another file name; it is a shortcut that points to the original file path. If the original file is deleted, the soft link becomes “broken”.

Q3. What is a Kernel in Linux?
- The Kernel is the core part of the Linux operating system that manages hardware resources and allows communication between hardware and software. It handles memory, processes, device management, and system calls.

Q5. What is the 'grep' command used for in Linux?
- The grep command is used to search for text patterns within files. It prints lines matching a given pattern.
Example: grep 'pattern' filename

Q6. Step1: Create user p1
- Step2: He should be part of 3 groups g1,g2,g3.
- Step3: whenever he creates a file automatically in the group section of file grp g1 should come.
---
- Create user p1: useradd p1
- Add p1 to groups g1, g2, g3: usermod -aG g1,g2,g3 p1
- Set g1 as the default group so any new files by p1 have group ownership as g1: usermod -g g1 p1

Q7. Step1: Create directory /tmp/bg as root user and create files inside it.
 - Step2: “abhi” should be the owner of the directory. He should be able to create files and delete files inside the directory and also he should be able to add content to all files inside the directory.

---
To create /tmp/bg as root user:
- mkdir /tmp/bg

Create some test files:
- touch /tmp/bg/file1.txt /tmp/bg/file2.txt

Change the ownership of the directory and files to the user “abhi”:
- chown -R abhi /tmp/bg

Set permissions so abhi can do anything in this directory (read, write, delete, add content):
- chmod -R 700 /tmp/bg



Note: The -R (recursive) flag makes sure both the directory and all files inside become abhi's property.

Q8. You suspect that a particular process is consuming excessive CPU resources on your Linux server. How would you identify and terminate this process?


---


Open up a tool called top in the terminal:
- top

This shows all running processes and their CPU usage.

Find the bad actor—the process using too much CPU (it’ll be at the top).

Note its PID (process ID), then type:
- kill <PID>

If it refuses to go away, use:
- kill -9 <PID>

That’s the force quit option!
