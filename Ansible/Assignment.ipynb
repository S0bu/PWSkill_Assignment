```
1. I installed Ansible on an Ubuntu control node using the Ansible PPA and `apt`.
2. I configured an inventory file with one managed host and SSH access.
3. I wrote and executed a basic Ansible playbook (`setup_web.yml`) that:
   - Updates the apt cache
   - Installs the `nginx` package
   - Ensures the `nginx` service is started and enabled at boot
4. The playbook ran successfully, and the managed node now has Nginx installed and running.
```
```
setup_web.yml:
---
- name: Basic web server setup
  hosts: web
  become: yes

  tasks:
    - name: Ensure apt cache is up to date
      apt:
        update_cache: yes

    - name: Install Nginx
      apt:
        name: nginx
        state: present

    - name: Ensure Nginx is running and enabled
      service:
        name: nginx
        state: started
        enabled: yes
```
