1. Explain the key features of Python that make it a popular choice for programming.
```
Key features of Python:
*It has simple and readable syntax close to English, which makes it easy to learn and maintain.

*It is interpreted and dynamically typed, so you can run code line by line and do not need explicit type declarations.

*It has a large standard library and rich ecosystem of third‑party packages (NumPy, Pandas, Django, etc.).

*It is portable and cross‑platform, running on Windows, macOS, Linux without code changes in most cases.

*It supports multiple paradigms: procedural, object‑oriented, and some functional programming features (lambda, map, filter).
```
2. Describe the role of predefined keywords in Python and provide examples of how they are used in a program.
```
Predefined keywords are reserved words that have special meaning in the language and cannot be used as variable names.

Examples:
if, elif, else – for conditional branching.
for, while, break, continue – for loops and loop control.
def, class, return – for functions and classes.
True, False, None – special constant values.

Example program:
def check_number(x):
    if x > 0:
        return "Positive"
    elif x == 0:
        return "Zero"
    else:
        return "Negative"
result = check_number(5)
print(result)

Keywords: def, if, elif, else, and return.
```
3. Compare and contrast mutable and immutable objects in Python with examples.
```
Mutable vs immutable objects:
*Mutable objects can be changed in place after creation.
*Immutable objects cannot be changed; any “modification” creates a new object.

Common examples:
Mutable: list, dict, set.
Immutable: int, float, str, tuple.

Example snippets:
# Immutable
a = 10
b = a
a = a + 5     
print(a, b)    

# Mutable
lst1 = [1, 2, 3]
lst2 = lst1
lst1.append(4)
print(lst1, lst2)
```
4. Discuss the different types of operators in Python and provide examples of how they are used.
```
Types of operators in Python:
* Arithmetic: + - * / // % **
  Example of use case:
  x = 5 + 3      
  y = 10 // 3    
  z = 2 ** 3

* Logical: and or not
  Example of use case:
  print(a > 3 and a < 10)

* Relational (comparison): == != > < >= <=
  Example of use case:
  a = 5
  print(a > 3)

* Assignment: = += -= *= /= //= %= **=
  Example of use case:
  x = 5
  x += 2

* Bitwise: & | ^ ~ << >>
  Example of use case:
  print(5 & 3)

* Membership: in, not in
  Example of use case:
  print(3 in [1, 2, 3])

* Identity: is, is not
  Example of use case:
  a = [1, 2]
  b = a
  print(a is b)
```
5. Explain the concept of type casting in Python with examples.
```
Type casting converts a value from one data type to another using built‑in functions.

Common examples:
# int to float
a = 5
b = float(a)     # 5.0

# string to int
s = "123"
n = int(s)       # 123

# int to string
x = 10
s2 = str(x)      # "10"

# list to tuple
lst = [1, 2, 3]
t = tuple(lst)   # (1, 2, 3)

# string to list of characters
name = "abc"
chars = list(name)  # ['a', 'b', 'c']
```
6. How do conditional statements work in Python? Illustrate with examples.
```
Conditional statements execute different blocks of code based on Boolean expressions.

Example:
x = 10
if x > 0:
    print("Positive")
elif x == 0:
    print("Zero")
else:
    print("Negative")

We can also use nested if or short single‑line conditionals:

Example:
age = 18
status = "Adult" if age >= 18 else "Minor"
print(status)
```
7. Describe the different types of loops in Python and their use cases with examples.
```
Main loop types are 'for' and 'while'.

'for' loop: iterates over a sequence (list, string, range, etc.).
Example:
for i in range(1, 6):
    print(i)

'while' loop: repeats as long as a condition is True.
Example:
count = 1
while count <= 5:
    print(count)
    count += 1
```
